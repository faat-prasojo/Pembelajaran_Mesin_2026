### Kegunaan dari setipa library tersebut yaitu

- PyPREP: digunakan untuk prapemrosesan data EEG (Elektroensefalogram / rekaman aktivitas listrik otak). Kegunaan utamanya adalah membersihkan data mentah EEG secara otomatis dari noise (gangguan/artefak) dan memperbaiki saluran (channel) sinyal yang buruk sebelum data tersebut dimasukkan ke dalam model analisis.

- Scipy (SciPy): digunakan untuk komputasi matematika dan ilmiah di Python. Dalam konteks sinyal seperti otak dan jantung, SciPy hampir selalu digunakan untuk pemrosesan sinyal (signal processing). Misalnya, untuk membuat filter frekuensi yang membuang gangguan dengung listrik pada rekaman medis, atau menghitung transformasi matematis dari sinyal tersebut.

- wandp (Kemungkinan besar salah ketik dari wandb): Fungsinya adalah untuk melacak eksperimen (tracking). Saat model AI dilatih, wandb akan membuat grafik metrik (seperti akurasi dan tingkat kesalahan/loss) secara real-time ke dasbor online agar Anda bisa memantau dan membandingkan performa berbagai model AI yang sedang Anda buat.

- epECG: digunakan untuk pemrosesan data EKG / ECG (Elektrokardiogram / rekaman aktivitas jantung). Fungsinya mencakup membaca format rekaman medis jantung, membersihkan sinyal jantung, hingga mengekstrak fitur penting (seperti mendeteksi jarak antar detak/gelombang QRS) untuk dianalisis lebih lanjut.
