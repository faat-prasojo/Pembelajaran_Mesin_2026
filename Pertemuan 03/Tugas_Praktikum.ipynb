```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Membaca data
df = pd.read_csv('wbc.csv')
```

### 1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
# Membuang kolom id dan Unnamed: 32 (jika ada)
df_clean = df.drop(columns=['id', 'Unnamed: 32'], errors='ignore')

# Cek hasil penghapusan
print(df_clean.columns)
```

### 2. Lakukan proses encoding pada kolom "diagnosis".

```python
# Inisialisasi LabelEncoder
le = LabelEncoder()

# Melakukan encoding
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

# Cek hasil encoding (M=1, B=0)
print(df_clean['diagnosis'].head())
```

### 3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.

```python
# Memisahkan target (y) dan fitur (X)
X = df_clean.drop('diagnosis', axis=1)
y = df_clean['diagnosis']

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Melakukan standarisasi pada fitur (X)
X_scaled = scaler.fit_transform(X)

# Mengembalikan bentuk array numpy menjadi DataFrame pandas agar rapi
X_final = pd.DataFrame(X_scaled, columns=X.columns)

# Cek hasil standarisasi
print(X_final.head())
```

### 4. Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.

```python
# Membagi data dengan test_size=0.20 (20%) dan proporsi kelas yang sama (stratify)
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y,
    test_size=0.2,
    stratify=y,
    random_state=42 # random_state agar hasil pembagian selalu sama saat di-run ulang
)

# Mengecek dimensi data hasil split
print("Ukuran data latih (X_train):", X_train.shape)
print("Ukuran data uji (X_test):", X_test.shape)
```
