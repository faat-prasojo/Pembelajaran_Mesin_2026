# Tugas Praktikum

1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
```

```python
# Dataset Awal
print("Dimensi data : ", df.shape)
display(df.head())
```

```python
# Variabel yang tidak digunakan (id dan Unnamed)
unusable_cols = ['id', 'Unnamed: 32']
df_cleaned = df.drop(columns=unusable_cols)
```

```python
print("Kolom setelah membuang variabel yang tidak digunakan:")
print(df_cleaned.columns.tolist())
```

2. Lakukan proses encoding pada kolom "diagnosis".

```python
le = LabelEncoder()

# encoding pada kolom diagnosis
df_cleaned['diagnosis'] = le.fit_transform(df_cleaned['diagnosis'])

print("Mapping kelas:", dict(zip(le.classes_, range(len(le.classes_)))))
display(df_cleaned[['diagnosis']].head())
```

3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
# Kolom numerik (selain kolom diagnosis)
feature_cols = [col for col in df_cleaned.columns if col != 'diagnosis']
```

```python
scaler = StandardScaler()

# Standardisasi pada kolom fitur
df_cleaned[feature_cols] = scaler.fit_transform(df_cleaned[feature_cols])

# Hasil akhir data
display(df_cleaned.head())
```
